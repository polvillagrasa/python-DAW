__Exercici 1:__
 
Donada una taula de n elements enters introduïts per l’usuari, fer un algorisme que ens indiqui quants d’aquests són negatius i quants no. El zero es considera neutre.

__Exercici 2:__ 

Escriure un algorisme que permeti calcular el quadrat dels 100 primer números enters i a continuació els vagi guardant en una taula.	

__Exercici 3:__

Realitzar un algorisme que ens permeti entrar n nombres enters en una taula i a continuació que inverteixi la taula (el primer element passa a l’últim introduït, el segon al penúltim,...). Una vegada invertida la taula mostra-la per pantalla.	

__Exercici 4:__
    
Realitzar un algorisme que donada una taula de N elements, intercanvïi els elements de dos en dos. Si la taula té un nombre imparell d’elements, l’últim no es canvia amb ningú. Finalment, mostrar per pantalla la taula modificada.

__Exercici 5:__ 
    
Donada una taula de n elements d’enters introduïts per l’usuari, suma tots els elements de la taula fent servir una funció de la llibreria "numpy".	

__Exercici 6:__

Tenim N temperatures guardades en una taula. Calcula la mitjana mitjançant una funció de la llibreria "numpy" desrés digues quin és el número de temperatures més grans o iguals a la mitjana.	

__Exercici 7:__ 
    
Demana una cadena de caràcters. Després digues quines lletres de l'abecedari s'han introduït.

__Exercici 8:__
    
Demana una cadena de caràcters. Després digues per cada lletra de l'abecedari quantes vegades apareix. (‘a’-‘z’)

__Exercici 9:__

Es necessita un aplicatiu on es pugui introduir el codi postal d’habitants de 20 poblacions catalanes. Tindrem 1000 habitants. Es calcularà el valor de forma aleatòria per cada habitant enregistrat. Finalment es demana que es digui el nom de la població amb més habitants i el número d’habitants que té, així com la mitja d’habitants de les 20 poblacions. Les poblacions s’anomenen “Població1”, “Població2”, “Població3”, etc.
