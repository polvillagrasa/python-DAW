__Exemple 2__: Àmbit de les variables

Les variables poden ser globals o locals, segons on es creïn. Per saber si les variables es modificaran sense retornar-les, també ens afecta el seu tipus, ja que les variables de tipus simples es passen per valor (es crea una còpia local automàticament) i, en canvi, les de tipus compost es passen per referència (es modifica la variable original).

__1.__ Puc fer servir el valor d'una variable global:

```python
def imprimir():
    print(valor)

valor=0
imprimir()    
```

__2.__ Però NO puc modificar-la.

```python
def incrementar():
    valor+=1
    print(valor)

valor=0
incrementar()    
```

__3.__ Per poder-ho fer, he d'indicar que busqui la variable global.

```python
def incrementar():
    global valor
    valor+=1
    print(valor)

valor=0
incrementar()    
```

4. El __millor__, tret que sigui una variable realment GLOBAL, és __passar les variables per paràmetre__ i retornar-les.

__Exemple 3: Pas per valor__

En les variables de tipus simple, les variables passades per paràmetre es passen __per valor__.

__a.__ Si no reassignem el valor retornat, la variable no es modifica.

```python
def incrementar(valor):
    valor+=1
    return valor

valor = 0
incrementar(valor) #Falta l'assignació (valor = incrementar(valor))
print(valor)
```

__b.__ Aquí, en canvi, si es modifica, ja que fem la reassignació.

```python
def incrementar(valor):
    valor+=1
    
    return valor

valor = 0
valor = incrementar(valor)
print(valor)    
```

__Exemple 4: Pas per referència__

Amb les taules, com el pas es fa per referència, es realitzen els canvis sobre la pròpia variable passada.

__a.__ En aquest cas, a la funció incrementarTaula no retornem un valor, però com treballem sobre la taula proporcionada, els canvis es mantenen.

```python
def incrementarTaula(taula):
    for i in range(len(taula)):
        taula[i]+=1

nums = [0,1]        
print(nums)
incrementarTaula(nums)
print(nums)
```

__b.__ Si no volem alterar la taula passada (de manera que tinguem dues taules) necessitem crear una còpia.

```python
def incrementarTaula(taula):
    taulaAux = taula.copy()
    for i in range(len(taulaAux)):
        taulaAux[i]+=1
    
    return taulaAux

nums = [0,1]        
nums2=incrementarTaula(nums)
print(nums) #La taula original no es modifica
print(nums2) #La taula retornada té els canvis
```

Vull fer un programa per invertir una taula. Necessito imprimir la taula diverses vegades.

* Quan treballavem sense funcions, primer definiem les dimensions de la taula i després feiem servir aquestes variables a tot el codi.
* En encapsular el codi en funcions, ens cal trobar les dimensions dins de la funció, per tal d'estalviar-nos passar-ho com a paràmetre.
* Com la taula que volem invertir no ens fa falta tenir-la en una nova taula, podem fer les funcions __sense *return*__.

```python
def demanarValors(taula): #Procediment per omplir una taula
    n = taula.shape[0]
    m = taula.shape[1]
    for fila in range(n):      
        for col in range(m):
            taula[fila,col] = int(input("Introdueix un valor:"))

def imprimirTaula(taula): #Procediment per imprimir una taula
    n = taula.shape[0]
    m = taula.shape[1]
    for i in range (n):
        for j in range(m):
            print("--",end="")
            
    for i in range (n):
        print("\n|",end="")
        for j in range(m):
            valor = taula[i,j]
            print(valor,end=" ")
    
    print()        
    for i in range (n):
        for j in range(m):
            print("--",end="")
    print()
    
def invertirTaula(taula): #Procediment per invertir la taula
    n = taula.shape[0]
    m = taula.shape[1]
    for fila in range(round(n/2)):    
        filaI = n-fila-1
        for col in range(m):
            colI = m-col-1
            aux = taula[fila,col]
            taula[fila,col] = taula[filaI,colI]
            taula[filaI,colI] = aux
```

```python
n = 2 #Quantitat de files
m = 5 #Quantitat de columnes
llistat = np.zeros((n,m))
imprimirTaula(llistat)
demanarValors(llistat)
imprimirTaula(llistat)
invertirTaula(llistat)
imprimirTaula(llistat) # Els canvis queden reflectits tot i que no hi hagi return
```

Ara, vull poder tenir la llista original i la llista invertida. Necessitaré modificar la funció d'invertir per tal de que no em modifiqui la taula original.

```python
def invertirTaula(taula): #Procediment per invertir la taula
    taulaIn = taula.copy() #Fem una còpia de la taula i treballem sobre aquesta
    n = taulaIn.shape[0]
    m = taulaIn.shape[1]
    for fila in range(round(n/2)):    
        filaI = n-fila-1
        for col in range(m):
            colI = m-col-1
            aux = taulaIn[fila,col]
            taulaIn[fila,col] = taulaIn[filaI,colI]
            taulaIn[filaI,colI] = aux
            
    return taulaIn #Retornem la còpia modificada
```

```python
n = 2 #Quantitat de files
m = 5 #Quantitat de columnes
llistat = np.zeros((n,m))
imprimirTaula(llistat)
demanarValors(llistat)
llistatInvertit = invertirTaula(llistat)
imprimirTaula(llistat) #Imprimim la taula original després de cridar a invertirTaula, però no hi ha canvis ja que el que invertim és una còpia
imprimirTaula(llistatInvertit) #Imprimim la taula retornada per invertirTaula, aquesta sí que està invertida
```
