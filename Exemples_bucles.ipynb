## BUCLES FOR

Els bucles __*for*__ ens permeten executar un mateix codi un nombre predeterminat de vegades.

Es poden fer servir de diverses maneres.

### __1. In range__

- Permet executar el codi N vegades.
- S'utilitza una variable (habitualment amb el nom __i__) que pren els valors del rang.
- El rang comença al 0 si només s'indica un valor. 
- En cas de voler començar amb un altre nombre, s'especifica el valor d'inici del rang.
- L'últim valor no s'inclou mai.

__Exemple 1:__ Executar un codi N vegades. 

Demanar  noms i saludar a cadascú:

```python
for i in range(3): # Per a la i en el rang [0-2]
    nom = input("Com et dius?")
    print(f"Hola {nom}!")
```

__Exemple 2:__ Executar un codi N vegades, fent servir el valor de la *i*.

Demanar  noms i assignar el número:

```python
for i in range(3): # Per a la i en el rang [0-2]
    nom = input("Com et dius?")
    print(f"Hola {nom}! Et correspon el número {i+1}")
```

__Exemple 3:__ Executar un codi N vegades, fent servir el valor de la *i* començant directament per l'1.

Demanar  noms i assignar el número: 

```python
for i in range(1, 4): # Per a la i en el rang [0-3)
    nom = input("Com et dius?")
    print(f"Hola {nom}! Et correspon el número {i}")
```

__Exemple 4:__ Iterar en un rang concret, fent increments de N a la variable.

Volem saber els nombres parells entre el 10 i el 20 (sense incloure aquest últim):

```python
for i in range (10,20,2):
    print(i)
```

__Exemple 5:__ També podem iterar enrere.

Volem fer un compte enrere del 100 al 0, de 10 en 10:

```python
for i in range (100,-1,-10):
    print(i)
```

__Exemple exercici 15__

Tres amics que comparteixen pis volen veure si estan gastant tots els mateixos diners en menjar.  S’ha de desenvolupar un programa que permeti fer aquests càlculs de manera ràpida. El programa els hi demanarà el nom, el nombre de coses comprades i,  després, el valor de cadascuna de les coses que han comprat. 

El programa ha de dir:

- Quantes coses ha comprat cadascú.
- Quant ha gastat en total cadascú.
- Quant ha gastat de mitjana cadascú (preu/cosa).
- Qui ha gastat menys i més en total.

```python
mes_gastat = -float("inf")
menys_gastat = float("inf")
for i in range(3):
    total = 0
    nom = input(f"Nom de l'amic {i+1}: ")
    num_coses = int(input(f"Nombre de coses comprades per {nom}: "))
    print(f" {nom} ha comprat {num_coses} coses")
    for j in range(num_coses):
        preu_cosa = float(input(f"Preu de cada cosa comprada per {nom}: "))
        total += preu_cosa
    mitjana = total / num_coses
    if(total > mes_gastat):
        nom_mes_gastat = nom
        mes_gastat = total
    if(total < menys_gastat):
        nom_menys_gastat = nom
        menys_gastat = total
    print(total)
    print(mitjana)
print(nom_mes_gastat)
print(nom_menys_gastat)
```

### __2. Exemples de *for* amb strings i llistes:__

Treballar amb strings i amb llistes és molt semblant. Els fors ens permeten treballar amb ells de dues maneres:
1. Fent servir in range. L'avantatge és que ens permet accedir de nou a la posició, per exemple per modificar el valor.
2. Fent referència a l'element en cas de només voler consultar-lo o guardar-lo a un lloc diferent.

__Exemple 1:__ Volem trobar el nombre de lletres "a".

1. Fem referència als elements, ja que no necessitem accedir a la posició del string per modificar-lo.

```python
cadena = input("Introdueix una paraula o frase: ")
num = 0
for i in cadena:
    if i.lower() == "a":
        num+=1

print(f"A la cadena '{cadena}' hi ha {num} lletres A")
```

__Exemple 2:__ Volem saber els valors guardats a una llista.

1. Fem referència als elements, ja que no necessitem accedir posteriorment al llistat.

```python
llista = [3,4,1,9,8,7,56,0]

for num in llista:
    print(num)
```

__Exemple 3:__ Volem duplicar els valors guardats a una llista.

1. Fem servir el valor de posició del llistat, ja que podem fer servir l'índex per canviar el valor.

```python
llista = [3,4,1,9,8,7,56,0]
n = len(llista)

for i in range (n): #For utilitzant el valor de posició del llistat
    llista[i] = llista[i]*2
    print(llista[i])
```

__Exemple 4:__ Volem saber els últims 3 nombres d'un llistat amb longitud N.

1. Igual que en el cas anterior, farem servir l'índex si ens cal canviar el valor posteriorment.

```python
llista = [3,4,1,9,8,7,56,0]
n = len(llista)

for i in range (n-3,n): #For utilitzant el valor de posició del llistat
    print(llista[i])
```

En aquest cas, si volem només imprimir els tres últims valors, farem "slicing". Això permet treballar en un moment determinat amb una part de la llista, però sense modificar la original.

```python
llista = [3,4,1,9,8,7,56,0]
n = len(llista)

for num in llista[n-3:]: #For realitzant slicing
    print(num)
```

__Exemple 5:__ Volem fer un programa que, donades dues llistes, una amb noms i l'altra amb l'aportació feta, digui a cada persona quant li falta per arribar a 20€.

```python
noms = ['Joan','Julia','Pep','Claudia']
diners = [3,4,23,9]

for nom, aportacio in zip(noms,diners):
    if aportacio<20:
        print(f"{nom} et falten {20-aportacio}€ per arribar a 20€ d'aportació.")
    else: 
        print(f"{nom} ja has fet l'aportació corresponent!")
```

## __2. While__

- Permet executar el codi un nombre indeterminat de vegades
- S'estableix una condició mentre la qual el codi s'executa
- Cal predefinir el valor de les variables emprades en la condició.

__Exemple 1:__
    
Demanar lletres fins que l'usuari introdueixi un punt. El programa comptarà el nombre de lletres _A_ que conté la cadena de caràcters introduïda.

```python
lletra = input("Introdueix una lletra, . per acabar d'introduir: ")
num = 0
cadena = ""
while lletra != ".":
    cadena +=lletra
    if lletra == "a" or lletra == "A":
       num+=1
    
    lletra = input("Introdueix una lletra, 0 per acabar d'introduir: ")
       
    
print(f"A la cadena '{cadena}' hi ha {num} lletres A")
```

__Exemple 2:__ Executar un codi fins que l'usuari encerti.

```python
import random
num = random.randint(0,100)
num2 = random.randint(0,100)

print(f"Quant dona la suma {num} + {num2}?")

encertat = False
while(not encertat):
    suma = float(input("Quant dona la suma {num} + {num2}?"))
    if (num+num2) != suma:
        print("Incorrecte!")
    else:
        print("Correcte!")
        encertat=True
```
