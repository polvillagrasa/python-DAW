<h4> 1. Classes i objectes</h4>

```python
class Cotxe:
    # Constructor
    def __init__(self, color, num_portes, num_places, descapotable, consum, propietari):
        self.color = color
        self.num_portes = num_portes
        self.num_places = num_places
        self.descapotable = descapotable
        self.consum = consum
        self.__propietari = propietari #atribut privat

    # Getter per obtenir el color
    def get_color(self):
        return self.color

    # Setter per establir el color
    def set_color(self, color):
        self.color = color
        
    # Getter per obtenir el propietari
    def get_propietari(self):
        return self.__propietari

    # Setter per establir el propietari
    def set_propietari(self, propietari):
        self.__propietari = propietari
        
    # Getter per obtenir el número de portes
    def get_num_portes(self):
        return self.num_portes

    # Setter per establir el número de portes
    def set_num_portes(self, num_portes):
        self.num_portes = num_portes
    
    # Getter per obtenir el número de places
    def get_num_places(self):
        return self.num_places
    
    # Setter per establir el número de places
    def set_num_places(self, num_places):
        self.num_places = num_places
        
    # Getter per obtenir si és descapotable
    def descapotable(self):
        return self.descapotable

    # Setter per establir si és descapotable
    def descapotable(self, descapotable):
        self.descapotable = descapotable
        
    def calcular_despesa(self, preu_combustible):
        return self.consum*preu_combustible
        
        
    def __str__(self):    
         return f'''Cotxe:
        Color: {self.color}
        Portes: {self.num_portes}
        Places: {self.num_places}
        Descapotable: {'Sí' if self.descapotable else 'No'}
        Propietari: {self.__propietari}'''

    # Altres getters i setters...
```

```python
# Crear un objecte de la classe Cotxe
cotxe1 = Cotxe(color="Blau", num_portes=4, num_places=5, descapotable=False, propietari="Joan")

# Accés i modificació de l'atribut propietari
print(f"Propietari del cotxe: {cotxe1.get_propietari()}")
cotxe1.propietari="Pere"
print(f"Propietari del cotxe: {cotxe1.get_propietari()}")
cotxe1.set_propietari("Pere")
print(f"Nou propietari del cotxe: {cotxe1.get_propietari()}")

```

<h4>2. Herència </h4>

__Classe Base Vehicle de motor__

* La classe __Vehicle_motor__ ens permetrà tenir una classe base que simplifiqui la creació d'altres tipus de vehicles com cotxes, motos, etcètera.
* Hem canviat la visibilitat de "propietari" a __protected__, de manera que les classes filles també puguin accedir-hi fàcilment.

```python
class Vehicle_motor:
    # Constructor
    def __init__(self, color, num_rodes, velocitat, motor, consum, propietari):
        self.color = color #string
        self.num_rodes = num_rodes #int
        self.velocitat = velocitat #float
        self.motor = motor #boolean ences/apagat
        self.consum = consum #float
        self._propietari = propietari #atribut protegit, string

    # Getter per obtenir el color
    def get_color(self):
        return self.color

    # Setter per establir el color
    def set_color(self, color):
        self.color = color
        
    # Getter per obtenir el propietari
    def get_propietari(self):
        return self._propietari

    # Setter per establir el propietari
    def set_propietari(self, propietari):
        self._propietari = propietari
        
    #Mètode per calcular la despesa en € per cada 100km    
    def calcular_despesa(self, preu_combustible):
        return self.consum*preu_combustible
    
    #Mètode per encendre motor
    def encendreMotor(self):
        self.motor = True

    #Mètode per apagar motor
    def apagarMotor(self):
        self.motor = False

    #Mètode per accelerar
    def accelerar(self, valor):
        self.velocitat +=valor

    #Mètode per frenar
    def frenar(self, valor):
        self.velocitat -=valor
        
    #Mètode per imprimir les dades d'un vehicle
    def __str__(self):    
         return f'''Vehicle:
        Color: {self.color}
        Nº de rodes: {self.num_rodes}
        Consum: {self.consum}
        Propietari: {self._propietari}'''

    # Altres getters i setters...
```

```python
class Gestor_Vehicles:

    def demanar_dades_vehicle(self):
        propietari = input("Com et dius? ")
        color = input("De quin color és el vehicle? ")
        num_rodes = int(input("Quantes rodes té? "))
        consum = int(input("Quin és el seu consum? litres/100km"))

        vehicle = Vehicle_motor(color, num_rodes, 0, False, consum, propietari) #passem directament la velocitat i el motor perquè són dades de moviment, 
                                                                                #es podria definir al propi constructor i eliminar-los dels paràmetres
        return vehicle

# Codi principal
if __name__ == "__main__":
    gestor = Gestor_Vehicles()
    vehicles = []
    continuar = input("Vols afegir un nou vehicle? S/N").upper()
    while (continuar=="S"):
        nou_vehicle = gestor.demanar_dades_vehicle()
        vehicles.append(nou_vehicle)
        print(f"S'ha afegit un nou vehicle (nom del propietari: {nou_vehicle.get_propietari()}).")
        continuar = input("Vols afegir un nou vehicle? S/N").upper()
```

```python
for v in vehicles:
    print(v)
```

__Classe Derivada Cotxe__

* La classe __Cotxe__ hereta tots els atributs de __Vehicle_motor__ així com els atributs. No cal, per tant, tornar a definir-los.
* Cridem al constructor de vehicle amb __super()__ per passar-li els valors dels atributs en instanciar l'objecte.
* Els nous atributs sí que els hem d'assignar manualment.
* Es poden tornar a implementar els mètodes, això es coneix com a __sobreescriptura__. És el cas del mètode __str__.
* També podriem cridar al mètode __str__ de la classe base si volguéssim mantenir el que fa i simplement afegir-li coses.

```python
class Cotxe(Vehicle_motor):
    # Constructor
    def __init__(self, color, velocitat, motor, consum, propietari, num_portes, num_places, descapotable):
        super().__init__(color, 4, velocitat, motor, consum, propietari)
        self.num_portes = num_portes
        self.num_places = num_places
        self.descapotable = descapotable
        
    # Getter per obtenir el número de portes
    def get_num_portes(self):
        return self.num_portes

    # Setter per establir el número de portes
    def set_num_portes(self, num_portes):
        self.num_portes = num_portes
    
    # Getter per obtenir el número de places
    def get_num_places(self):
        return self.num_places
    
    # Setter per establir el número de places
    def set_num_places(self, num_places):
        self.num_places = num_places
        
    # Getter per obtenir si és descapotable
    def descapotable(self):
        return self.descapotable

    # Setter per establir si és descapotable
    def descapotable(self, descapotable):
        self.descapotable = descapotable
        
    #Sobreescriptura del mètode __str__ de Vehicle_motor
    def __str__(self):
        return f'''Cotxe:
        Color: {self.color}
        Portes: {self.num_portes}
        Places: {self.num_places}
        Nº de rodes: {self.num_rodes}
        Consum: {self.consum}
        Descapotable: {'Sí' if self.descapotable else 'No'}
        Propietari: {self._propietari}'''
    
    #Sobreescriptura del mètode __str__ de Vehicle_motor amb crida al mètode heretat
    #def __str__(self):
    #    return super().__str__()+f'''
    #    Places: {self.num_places}
    #    Portes: {self.num_portes}
    #    Descapotable: {'Sí' if self.descapotable else 'No'}'''
```
