__Exercici 1:__
 
    Calcular l’àrea d’un cercle. Fórmula : pi*r2.


__Exercici 2:__ 

    Introduir dos números i calcular la seva suma.

__Exercici 3:__

    Introduir un import i calcular què ha de pagar si li fem 20% de descompte.

__Exercici 4:__
    
    Introduir un import i un descompte en tant per cent. Calcular l’import  net.


__Exercici 5:__ 
    
    Demanar el nom de l’usuari per teclat i que l’ordinador el saludi a posteriori.

__Exercici 6:__ 
    
    Fes un programet que demani la nota de tres exàmens i ens faci la mitjana resultant. Les qualificacions poden tenir decimals.

__Exercici 7:__
    
    Introduir els anys d’una persona i obtenir l’ equivalent en dies i segons.

__Exercici 8:__

    Intercanviar els valors de dues variables prèviament introduïdes.


__Exercici 9:__

    Si tenim tres variables A,B i C. Intercanviar els valors de la següent manera:
        
        - B pren del valor de A 
        - C pren el valor de B
        - A pren el valor de C.
    
    Només podem utilitzar una variable auxiliar.


__Exercici 10:__
    
    Realitzar un algorisme que calculi l’àrea d’un triangle

__Exercici 11:__
    
    Genereu dos nombres aleatoris enters del 0 al 100. Mostreu-los per pantalla i pregunteu quin és el resultat de la seva suma. L’usuari ha de respondre amb un nombre, i li haureu d’indicar si aquest és correcte (true) o incorrecte (false).


__Exercici 12:__

    Comproveu si un alumne aprova l'assignatura, tenint en compte el següent:
    la nota mínima que podem treure als exàmens per fer mitja és un 3,5 i  conten com un 60% de la nota final.


__Exercici 13:__

    A partir d’un valor X , calcular el valor d’ aquest polinomi X^3 + 3X^2 + 7X +7.

__Exercici 14:__

    Demanar un número enter que signifiqui una quantitat d’euros i digui quants bitllets de 5, 10, 20,50 i 100 son equivalents.
    Exemple:
        255 són:
        51 bitllets de 5 i me'n sobren 0
        25 bitllets de 10 i me'n sobren 5
        12 bitllets de 20  i me'n sobren 15
        5 bitllets de 50 i me'n sobren 5
        2 bitllets de 100 i me'n sobren 55


__Exercici 15:__

    Demanar un número enter que signifiqui una quantitat d’euros i digui la descomposició en bitllets de 5,10,20,50 i 100.
    Exemple:
        255 són:
        2 bitllets de 100
        1 bitllet de 50
        0 bitllets de 20
        0 bitllets de 10
        1 bitllets de 5
