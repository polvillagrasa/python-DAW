```python
class Cotxe:
    # Constructor
    def __init__(self, color, num_portes, num_places, descapotable, consum, propietari):
        self.color = color
        self.num_portes = num_portes
        self.num_places = num_places
        self.descapotable = descapotable
        self.consum = consum
        self.__propietari = propietari

    # Getter per obtenir el color
    def get_color(self):
        return self.color

    # Setter per establir el color
    def set_color(self, color):
        self.color = color
        
    # Getter per obtenir el propietari
    def get_propietari(self):
        return self.__propietari

    # Setter per establir el propietari
    def set_propietari(self, propietari):
        self.__propietari = propietari
        
    # Getter per obtenir el número de portes
    def get_num_portes(self):
        return self.num_portes

    # Setter per establir el número de portes
    def set_num_portes(self, num_portes):
        self.num_portes = num_portes
        
    # Getter per obtenir si és descapotable
    def descapotable(self):
        return self.descapotable

    # Setter per establir si és descapotable
    def descapotable(self, descapotable):
        self.descapotable = descapotable
        
    def calcular_despesa(self, preu_combustible):
        return self.consum*preu_combustible
        
        
    def __str__(self):    
         return f'''Cotxe:
        Color: {self.color}
        Portes: {self.num_portes}
        Places: {self.num_places}
        Descapotable: {'Sí' if self.descapotable else 'No'}
        Propietari: {self.__propietari}'''

    # Altres getters i setters...
```

```python
# Crear un objecte de la classe Cotxe
cotxe1 = Cotxe(color="Blau", num_portes=4, num_places=5, descapotable=False, consum=20, propietari="Joan")

# Accés i modificació de l'atribut propietari
print(f"Propietari del cotxe: {cotxe1.get_propietari()}")
cotxe1.__propietari="Pere"
print(f"Propietari del cotxe: {cotxe1.get_propietari()}")
cotxe1.set_propietari("Pere")
print(f"Nou propietari del cotxe: {cotxe1.get_propietari()}")

```

```python
def demanar_dades():
    propietari = input("Com et dius? ")
    color = input("De quin color és el cotxe? ")
    num_portes = int(input("Quantes portes té? "))
    num_places = int(input("Quantes places té? "))
    consum = int(input("Quin és el seu consum? litres/100km"))
    d = input("És descapotable? (S/N) ")

    descapotable = True if d.upper() == "S" else False

    cotxe = Cotxe(color, num_portes, num_places, descapotable, consum, propietari)
    return cotxe

# Codi principal
if __name__ == "__main__":
    cotxe1 = demanar_dades()
    print(cotxe1)
    print(f"El teu cotxe gasta {cotxe1.calcular_despesa(2)}€ per cada 100km")
    
```

```python
# Codi principal
if __name__ == "__main__":
    cotxes = []
    continuar = input("Vols afegir un nou cotxe? S/N").upper()
    while (continuar=="S"):
        nou_cotxe = demanar_dades()
        cotxes.append(nou_cotxe)
        print(f"S'ha afegit un nou cotxe (nom del propietari: {nou_cotxe.get_propietari()}).")
        continuar = input("Vols afegir un nou cotxe? S/N").upper()

        
```
